# nscf-расчёт

**nscf** (non-self-consistent field) — несамосогласованный расчёт: однократная диагонализация гамильтониана при **замороженной** зарядовой плотности, взятой из уже сошедшегося scf-расчёта. В отличие от обычного самосогласованного цикла, в nscf зарядовая плотность и потенциал **не обновляются** — они читаются из каталога `outdir` (у нас `./tmp_{prefix}`), куда их сохранил scf-расчёт, а для каждой k-точки выполняется одна диагонализация. Поэтому в выводе nscf нет строки `! total energy`: полная энергия здесь не вычисляется.

#### Вспомогательные функции

````python
DATA_DIR = '.'

Ha_TO_eV = 27.211386
Bohr_TO_A = 0.529177

def flt(el, path, default=0.0):
    node = el.find(path)
    t = node.text.strip() if node is not None and node.text else None
    return float(t) if t else default

def cell_from_xml(struct_el):
    return np.array([
        [float(x) * Bohr_TO_A for x in struct_el.find(f'.//cell/a{i}').text.split()]
        for i in range(1, 4)
    ])

def atoms_from_xml(struct_el):
    return [{'name': a.get('name'), 'coords': [float(x) for x in a.text.split()]}
            for a in struct_el.findall('.//atomic_positions/atom')]

def cell_to_a(cell):
    return np.linalg.norm(cell[0])

def make_structure(cell, atoms):
    species = [a['name'] for a in atoms]
    coords = [np.array(a['coords']) * Bohr_TO_A for a in atoms]
    return Structure(lattice=cell.tolist(), species=species,
                     coords=coords, coords_are_cartesian=True)

def parse_pw_in(text):
    params = {}
    for line in text.split('\n'):
        line = line.strip()
        if '=' in line and not line.startswith('!') and not line.startswith('&'):
            key, val = line.split('=', 1)
            params[key.strip()] = val.strip().rstrip(',')
    return params

def parse_scf(pw_out_path):
    ethrs, energies = [], []
    with open(pw_out_path) as f:
        for line in f:
            if line.strip().startswith('!'):
                continue
            m_ethr = re.search(r'ethr\s*=\s*([\d.E+-]+)', line)
            if m_ethr:
                ethrs.append(float(m_ethr.group(1)))
            m_e = re.search(r'total energy\s*=\s*([\-\d.]+)\s*Ry', line)
            if m_e:
                energies.append(float(m_e.group(1)))
    return energies, ethrs

def last_toten_ev(pw_out_path):
    last = None
    with open(pw_out_path) as f:
        for line in f:
            s = line.strip()
            if s.startswith('!') and 'total energy' in s:
                last = float(s.split('=')[1].split()[0]) * Ha_TO_eV / 2
    return last

def total_magnetization_from_pw_out(pw_out_path):
    last = 0.0
    with open(pw_out_path) as f:
        for line in f:
            if 'total magnetization' in line and 'Bohr mag/cell' in line:
                last = float(line.split('=')[1].split()[0])
    return last
````

## Ионный цикл

````python
def parse_ion_forces_from_pw_out(pw_out_path):
    blocks = []
    current = None
    with open(pw_out_path) as f:
        for line in f:
            s = line.strip()
            if s.startswith('Forces acting on atoms'):
                current = []
                blocks.append(current)
            elif current is not None and s.startswith('atom') and 'force =' in s:
                current.append(np.array([float(x) for x in s.split('force =')[1].split()]))
            elif current is not None and s.startswith('Total force'):
                current = None
    return [np.max(np.linalg.norm(np.array(b), axis=1)) * Ha_TO_eV / Bohr_TO_A
            for b in blocks if b]

data_ion_forces = []
for name in ['Ni', 'Al', 'Ni3Al']:
    f_max_list = parse_ion_forces_from_pw_out(f'{DATA_DIR}/{name}/result/{name}.pw.out')
    for i, f_max in enumerate(f_max_list):
        data_ion_forces.append({'Система': name, 'Ионный шаг': i, '|F_max| (эВ/Å)': f_max})

df_ion_forces = pd.DataFrame(data_ion_forces)
df_ion_forces
````

````python
def k_grid_from_pw_in(pw_text):
    lines = pw_text.split('\n')
    for i, line in enumerate(lines):
        if line.strip().upper().startswith('K_POINTS'):
            grid = lines[i + 1].strip().split()[:3]
            if len(grid) == 3 and all(x.isdigit() for x in grid):
                return tuple(int(x) for x in grid)
            return None
    return None

def k_grid_from_vasprun(vasprun_path):
    text = open(vasprun_path, encoding='utf-8', errors='replace').read()
    m = re.search(r'name="divisions">\s*([\d\s]+)<', text)
    return tuple(int(x) for x in m.group(1).split()[:3]) if m else None

Ni_kmesh_relax = k_grid_from_pw_in(Ni_pw_input_file)
Al_kmesh_relax = k_grid_from_pw_in(Al_pw_input_file)
Ni3Al_kmesh_relax = k_grid_from_pw_in(Ni3Al_pw_input_file)

Ni_kmesh_vasp = k_grid_from_vasprun('../VASP_run_03/Ni/result/vasprun.xml')
Al_kmesh_vasp = k_grid_from_vasprun('../VASP_run_03/Al/result/vasprun.xml')
Ni3Al_kmesh_vasp = k_grid_from_vasprun('../VASP_run_03/Ni3Al/result/vasprun.xml')

data_kpoints = {
    'Система': ['Ni', 'Al', 'Ni3Al'],
    'K-сетка QE (relax)': [f'{Ni_kmesh_relax[0]}×{Ni_kmesh_relax[1]}×{Ni_kmesh_relax[2]}',
              f'{Al_kmesh_relax[0]}×{Al_kmesh_relax[1]}×{Al_kmesh_relax[2]}',
              f'{Ni3Al_kmesh_relax[0]}×{Ni3Al_kmesh_relax[1]}×{Ni3Al_kmesh_relax[2]}'],
    'K-сетка QE (nscf)': [f'{Ni_kmesh[0]}×{Ni_kmesh[1]}×{Ni_kmesh[2]}',
              f'{Al_kmesh[0]}×{Al_kmesh[1]}×{Al_kmesh[2]}',
              f'{Ni3Al_kmesh[0]}×{Ni3Al_kmesh[1]}×{Ni3Al_kmesh[2]}'],
    'K-сетка VASP': [f'{Ni_kmesh_vasp[0]}×{Ni_kmesh_vasp[1]}×{Ni_kmesh_vasp[2]}',
              f'{Al_kmesh_vasp[0]}×{Al_kmesh_vasp[1]}×{Al_kmesh_vasp[2]}',
              f'{Ni3Al_kmesh_vasp[0]}×{Ni3Al_kmesh_vasp[1]}×{Ni3Al_kmesh_vasp[2]}'],
    'Высокосимм. точки': [Ni_hs_points, Al_hs_points, Ni3Al_hs_points],
}

df_kpoints = pd.DataFrame(data_kpoints)
df_kpoints
````

## Проектированный DOS (projwfc.x)

````python
def load_element_dos(system_dir, system_name):
    files = sorted(glob.glob(os.path.join(system_dir, f'{system_name}.pdos_atm#*')))
    element_files = {}
    for f in files:
        elem = re.search(r'atm#\d+\((\w+)\)', os.path.basename(f)).group(1)
        element_files.setdefault(elem, []).append(f)

    element_dos = {}
    for elem, elem_files in element_files.items():
        energy, up, dn = None, None, None
        for f in elem_files:
            data = np.loadtxt(f, skiprows=1)
            if energy is None:
                energy = data[:, 0]
                up = np.zeros_like(energy)
                dn = np.zeros_like(energy)
            n = data.shape[1]
            if n >= 5:   # spin-polarized: E, ldosup, ldosdw, [pdosup,pdosdw]*m (s:5,p:9,d:13 колонок).
            # Брать надо ldosup/ldosdw (сумма по всем m), иначе теряются 2/3 p и 4/5 d
                up += data[:, 1]; dn += data[:, 2]  # ldosup/ldosdw = сумма по всем m
            elif n == 3: # non-spin-polarized: E, ldos, pdos
                up += data[:, 2]
        element_dos[elem] = {'energy': energy, 'up': up, 'dn': dn}
    return element_dos
````

````python
Ni3Al_element_dos = load_element_dos(f'{DATA_DIR}/Ni3Al/result', 'Ni3Al')
Ni3Al_pdos_Al = Ni3Al_element_dos['Al']
Ni3Al_pdos_Ni = Ni3Al_element_dos['Ni']

# Раздельные спиновые каналы для тотального DOS (из pdos_tot: столбцы 1 и 2)
dens_up_total = Ni3Al_total_DOS_data[:, 1]
dens_down_total = Ni3Al_total_DOS_data[:, 2]

# PDOS каналы уже есть в словарях
dens_up_Al = Ni3Al_pdos_Al['up']
dens_down_Al = Ni3Al_pdos_Al['dn']
dens_up_Ni = Ni3Al_pdos_Ni['up']
dens_down_Ni = Ni3Al_pdos_Ni['dn']
````
